# LSFSurfaceIterative decomposition across five brightness quantiles with updated PALACE

This repository experiment complements `notebook_experiments_bspline_lsf.ipynb`.
It holds the fit implementation and PALACE bundle fixed at
`LSFSurfaceIterative / PALACE updated`, then runs the decomposition on five
real `FLUX_SCI` spectra selected across the brightness distribution of the
local median-stack FITS file.

Success criteria:

- keep the 12,401-element source wavelength grid exactly unchanged;
- verify the selected updated OH and diffuse PALACE tables by exact SHA256;
- prove that every `SkyDecomp` instance resolves those two updated files;
- select five fully finite spectra nearest the 5%, 27.5%, 50%, 72.5%, and 95%
  full-grid integrated-flux quantiles;
- complete all five default continuum → LSF → line refinement cycles;
- reproduce the full model from the returned components;
- verify the recently added O2 persistence contract;
- embed five native in-cell Plotly figures with observed flux, the refined full
  model, individual physical components, shifted residuals, and a robust
  residual envelope.


## Provenance and diagnostic contract

- Repository notebook: `skysub/notebook_experiments_lsf_surface_iterative_brightness.ipynb`.
- Companion experiment: `skysub/notebook_experiments_bspline_lsf.ipynb`
  performs the four-way `method × PALACE` comparison on three lunar regimes;
  this notebook isolates `LSFSurfaceIterative / PALACE updated` and spans five
  full-grid brightness quantiles.
- Code: Git commit `4acb1656c414b0835c472b131f8761d448d650cd` on `main`.
- Input: `skysub/lvmsframe_median_stack_1.2.1_every10.fits`, extension
  `FLUX_SCI`.
- PALACE root: `skysub/experiments_joint_diffuse_einstein_jax_v2/exports/`
  `baseline_compatible`.
- Updated OH input: `palace/PMD/pmd_popmodel_OH.dat`, SHA256
  `b0e2b763dc08c81b513fe51eed1505ff9982dcd89d975f8153e668e740ff5c65`.
- Updated diffuse input: `palace/PMD/pmd_refcont.dat`, SHA256
  `098bbe519170c491a2cbd1a862dec5e475eeb3300bbbeee252927cf8a36db961`.
- The shadow export supplies those two updated canonical files and links the
  unchanged PALACE support tables and solar spectrum back to the repository
  defaults. `SkyDecomp` receives this shadow root through `base_dir`, so its
  internal `palace/PMD` lookup uses the updated OH and diffuse tables.
- Brightness is used only for selecting rows. It is the trapezoidal integral of
  each observed spectrum over the complete source wavelength array.
- The model receives every selected wavelength sample. There is no binning,
  resampling, interpolation, smoothing, cropping, normalization, imputation,
  wavelength masking, or precision reduction.
- Flux is multiplied by the common factor `1e14`, matching the existing project
  decomposition example. This is a shared unit scaling, not a per-spectrum
  normalization.
- The stack has no inverse-variance extension. Every selected sample therefore
  receives unit weight. Reported chi-square-like values are diagnostic objective
  values, not physically calibrated chi-square statistics.
- The current model has Moon, `HO2`, `FeO`, `O2Ac`, OH, atomic, permitted O I,
  and molecular O2 components. It does not expose a separate zodiacal-light
  basis.
- The updated shadow export passed its loader/export contract, but the upstream
  50-spectrum comparison was worse than H11. These fits remain diagnostic-only;
  using the updated files here is not a production-acceptance claim.

Focused repository tests were also run at this commit: **13 passed and 16
failed**. Most failures are stale API/dataclass/FITS expectations after the new
`vector_o2`, `o2_prefit_amp`, and `VECTOR_O2` output contract; the final notebook
gate below is therefore an end-to-end diagnostic, not a clean unit-test-suite
acceptance claim.

```python
from dataclasses import asdict
from pathlib import Path
import hashlib
import os
import sys
import time

import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio
from astropy.io import fits
from IPython.display import display


cwd = Path.cwd().resolve()
root_candidates = [
    candidate
    for ancestor in (cwd, *cwd.parents)
    for candidate in (ancestor, ancestor / "lvmsky")
]
REPO_ROOT = next(
    path.resolve()
    for path in root_candidates
    if (path / "skysub" / "sky_decomp" / "fit.py").exists()
)
SKYSUB_ROOT = REPO_ROOT / "skysub"
SDSS5_ROOT = REPO_ROOT.parents[1]
UPDATED_PALACE_BASE_DIR = (
    SKYSUB_ROOT
    / "experiments_joint_diffuse_einstein_jax_v2"
    / "exports"
    / "baseline_compatible"
)
UPDATED_PALACE_FILES = {
    "OH": UPDATED_PALACE_BASE_DIR / "palace" / "PMD" / "pmd_popmodel_OH.dat",
    "Diffuse": UPDATED_PALACE_BASE_DIR / "palace" / "PMD" / "pmd_refcont.dat",
}
EXPECTED_UPDATED_PALACE_SHA256 = {
    "OH": "b0e2b763dc08c81b513fe51eed1505ff9982dcd89d975f8153e668e740ff5c65",
    "Diffuse": "098bbe519170c491a2cbd1a862dec5e475eeb3300bbbeee252927cf8a36db961",
}


def file_sha256(path):
    digest = hashlib.sha256()
    with path.open("rb") as stream:
        for chunk in iter(lambda: stream.read(1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()


assert UPDATED_PALACE_BASE_DIR.resolve() != SKYSUB_ROOT.resolve()
for palace_path in UPDATED_PALACE_FILES.values():
    assert palace_path.is_file(), palace_path
updated_palace_sha256 = {
    name: file_sha256(path) for name, path in UPDATED_PALACE_FILES.items()
}
assert updated_palace_sha256 == EXPECTED_UPDATED_PALACE_SHA256

# These fallbacks only satisfy the imported lvmdrp module's environment
# contract. SkyDecomp loads its scientific templates from the explicit updated
# PALACE shadow root below.
os.environ.setdefault("LVMCORE_DIR", str(SDSS5_ROOT / "lvmcore"))
os.environ.setdefault("LVM_SANDBOX", str(SDSS5_ROOT))
os.environ.setdefault("LVM_MASTER_DIR", str(SDSS5_ROOT))
sys.path.insert(0, str(SKYSUB_ROOT))

from sky_decomp.lsf_surface_iterative import (  # noqa: E402
    LSFSurfaceIterativeConfig,
    SkyDecompLSFSurfaceIterative,
)


pio.renderers.default = "plotly_mimetype"
EXPECTED_INTERPRETER = (
    "/opt/homebrew/Caskroom/miniconda/base/envs/lvmdrp_dev_311/bin/python"
)
assert Path(sys.executable).resolve() == Path(EXPECTED_INTERPRETER).resolve()

palace_provenance_table = pd.DataFrame(
    [
        {
            "component": component,
            "resolved_file": str(path.resolve()),
            "sha256": updated_palace_sha256[component],
        }
        for component, path in UPDATED_PALACE_FILES.items()
    ]
)

print(f"Python: {sys.executable}")
print(f"Repository root: {REPO_ROOT}")
print(f"Updated PALACE base_dir: {UPDATED_PALACE_BASE_DIR}")
for component, path in UPDATED_PALACE_FILES.items():
    print(f"{component} PALACE file: {path.resolve()}")
    print(f"{component} PALACE SHA256: {updated_palace_sha256[component]}")
print(f"Plotly renderer: {pio.renderers.default}")
with pd.option_context("display.max_colwidth", None):
    display(palace_provenance_table)
```

## Select five spectra on the unchanged source grid

```python
DATA_PATH = SKYSUB_ROOT / "lvmsframe_median_stack_1.2.1_every10.fits"
FLUX_EXTENSION = "FLUX_SCI"
FLUX_SCALE = 1.0e14
BRIGHTNESS_QUANTILES = np.array([0.05, 0.275, 0.50, 0.725, 0.95])

with fits.open(DATA_PATH, memmap=True) as hdul:
    source_wave = np.asarray(hdul["WAVE"].data).copy()
    wave = np.asarray(source_wave, dtype=np.float64)
    source_flux = np.asarray(hdul[FLUX_EXTENSION].data, dtype=np.float64)
    metadata = hdul["META"].data.copy()

assert wave.ndim == 1
assert wave.size == source_wave.size == 12_401
assert np.array_equal(wave, source_wave)
assert np.all(np.diff(wave) > 0.0)

fully_finite = np.all(np.isfinite(source_flux), axis=1)
selection_pool = np.flatnonzero(fully_finite)
brightness_integral = np.trapz(source_flux, wave, axis=1)
pool_brightness = brightness_integral[selection_pool]

selected_rows = []
for quantile in BRIGHTNESS_QUANTILES:
    target = np.quantile(pool_brightness, quantile)
    nearest = np.argmin(np.abs(pool_brightness - target))
    selected_rows.append(int(selection_pool[nearest]))

assert len(set(selected_rows)) == len(BRIGHTNESS_QUANTILES)

selected_flux = source_flux[selected_rows].copy() * FLUX_SCALE
selected_meta = metadata[selected_rows].copy()
assert selected_flux.shape == (5, wave.size)
assert np.all(np.isfinite(selected_flux))

wave_bytes = np.ascontiguousarray(wave.astype("<f8", copy=False)).tobytes()
wave_sha256 = hashlib.sha256(wave_bytes).hexdigest()

spectra = {}
selection_records = []
for rank, (quantile, row, flux, meta) in enumerate(
    zip(BRIGHTNESS_QUANTILES, selected_rows, selected_flux, selected_meta),
    start=1,
):
    label = f"Q{quantile * 100:g} brightness"
    integrated_scaled = float(brightness_integral[row] * FLUX_SCALE)
    spectra[label] = {
        "rank": rank,
        "quantile": float(quantile),
        "row": row,
        "flux": flux,
        "ivar": np.ones_like(wave),
        "meta": meta,
        "integrated_brightness": integrated_scaled,
    }
    selection_records.append(
        {
            "brightness_rank": rank,
            "target_quantile": float(quantile),
            "row": row,
            "expnum": int(meta["expnum"]),
            "mjd": int(meta["mjd"]),
            "integrated_flux_scaled_A": integrated_scaled,
            "median_flux_scaled": float(np.median(flux)),
            "moon_alt_deg": float(meta["moon_alt"]),
            "moon_fli": float(meta["moon_fli"]),
            "science_moon_sep_deg": float(meta["sci_moon_sep"]),
        }
    )

selection_table = pd.DataFrame(selection_records)
display(selection_table.style.format(precision=5))
print(
    f"Grid equality: exact; n={wave.size}; range={wave[0]:.6f}--{wave[-1]:.6f} A"
)
print(f"Grid SHA256 (float64 little-endian): {wave_sha256}")
print(
    f"Selection pool: {selection_pool.size}/{source_flux.shape[0]} fully finite rows; "
    "all selected model inputs retain every wavelength sample."
)

del source_flux
```

## Run the five-cycle iterative LSF decomposition with updated PALACE

```python
ITERATIVE_CONFIG = LSFSurfaceIterativeConfig()
MODEL_KWARGS = {
    "lsf_sigma": 0.5,
    "moon_smooth_lambda": 0.1,
    "moon_interline_boost": 0.0,
}

print("LSFSurfaceIterative configuration")
display(pd.Series(asdict(ITERATIVE_CONFIG), name="value").to_frame())
print(f"Model keyword arguments: {MODEL_KWARGS}")
print(f"PALACE base_dir passed to every fit: {UPDATED_PALACE_BASE_DIR}")
```

```python
def max_abs_difference(left, right):
    left = np.asarray(left, dtype=float)
    right = np.asarray(right, dtype=float)
    return float(np.max(np.abs(left - right)))


results = {}
runtime_state = {}
fit_records = []
stage_records = []

for label, spectrum in spectra.items():
    model = SkyDecompLSFSurfaceIterative(
        wave,
        base_dir=UPDATED_PALACE_BASE_DIR,
        config=ITERATIVE_CONFIG,
        **MODEL_KWARGS,
    )
    resolved_oh_path = model._pmd_path("pmd_popmodel_OH.dat").resolve()
    resolved_diffuse_path = model._pmd_path("pmd_refcont.dat").resolve()
    assert resolved_oh_path == UPDATED_PALACE_FILES["OH"].resolve()
    assert resolved_diffuse_path == UPDATED_PALACE_FILES["Diffuse"].resolve()
    started = time.perf_counter()
    result = model.fit(spectrum["flux"], spectrum["ivar"])
    wall_seconds = time.perf_counter() - started

    full_from_components = (
        result.components["moon"]
        + result.components["diffuse"]
        + result.components["oh"]
        + result.components["atom"]
        + result.components["orc"]
        + result.components["o2"]
    )
    diffuse_from_subcomponents = (
        result.components["ho2"]
        + result.components["feo"]
        + result.components["o2ac"]
    )
    o2_index = result.design_names.index("O2_b01")
    o2_from_persisted_template = result.coef[o2_index] * result.vector_o2

    model_closure = max_abs_difference(result.bestfit_lsf, full_from_components)
    diffuse_closure = max_abs_difference(
        result.components["diffuse"], diffuse_from_subcomponents
    )
    o2_closure = max_abs_difference(
        result.components["o2"], o2_from_persisted_template
    )

    np.testing.assert_allclose(
        result.bestfit_lsf, full_from_components, rtol=1.0e-11, atol=1.0e-11
    )
    np.testing.assert_allclose(
        result.components["diffuse"],
        diffuse_from_subcomponents,
        rtol=1.0e-11,
        atol=1.0e-11,
    )
    np.testing.assert_allclose(
        result.components["o2"],
        o2_from_persisted_template,
        rtol=1.0e-11,
        atol=1.0e-11,
    )

    results[label] = result
    runtime_state[label] = {
        "skyline_mask": model.skyline_mask.copy(),
        "stage_metrics": tuple(model.stage_metrics),
        "lsf_surface": model.lsf_surface.copy(),
    }
    fit_records.append(
        {
            "spectrum": label,
            "palace_inputs": "updated OH + diffuse",
            "row": spectrum["row"],
            "expnum": int(spectrum["meta"]["expnum"]),
            "integrated_flux_scaled_A": spectrum["integrated_brightness"],
            "fit_status": result.fit_status,
            "cycles": (
                f"{result.lsf_state.completed_cycles}/"
                f"{result.lsf_state.requested_cycles}"
            ),
            "wall_seconds": wall_seconds,
            "internal_fit_seconds": result.fit_elapsed_sec,
            "r2": result.r2,
            "rms_residual": result.rms_resid,
            "diagnostic_reduced_chi2": result.reduced_chi2,
            "o2_prefit_amp": result.o2_prefit_amp,
            "o2_final_coef": float(result.coef[o2_index]),
            "o2_component_peak": float(np.max(result.components["o2"])),
            "model_closure_max_abs": model_closure,
            "diffuse_closure_max_abs": diffuse_closure,
            "o2_persistence_max_abs": o2_closure,
        }
    )
    for metric in model.stage_metrics:
        stage_records.append({"spectrum": label, **metric})

    print(
        f"{label}: {result.fit_status}; "
        f"cycles={result.lsf_state.completed_cycles}/"
        f"{result.lsf_state.requested_cycles}; wall={wall_seconds:.2f} s"
    )

fit_table = pd.DataFrame(fit_records)
stage_table = pd.DataFrame(stage_records)
display(fit_table.style.format(precision=7))
```

## Native interactive Plotly component diagnostics

```python
def robust_mad(values):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    if values.size == 0:
        return np.nan
    center = np.median(values)
    return float(1.4826 * np.median(np.abs(values - center)))


COMPONENT_STYLES = {
    "moon": {"label": "Moon continuum", "color": "#e68a00", "dash": "dash"},
    "ho2": {"label": "HO2 diffuse", "color": "#00a6a6", "dash": "dash"},
    "feo": {"label": "FeO diffuse", "color": "#2ca02c", "dash": "dash"},
    "o2ac": {"label": "O2Ac diffuse", "color": "#9467bd", "dash": "dash"},
    "oh": {"label": "OH emission", "color": "#1f77b4", "dash": "dot"},
    "atom": {"label": "Atomic emission", "color": "#8c564b", "dash": "dot"},
    "orc": {"label": "Permitted O I emission", "color": "#e377c2", "dash": "dot"},
    "o2": {"label": "Molecular O2 emission", "color": "#17becf", "dash": "dot"},
}

WAVELENGTH_RANGES = {
    "Full": [float(wave[0]), float(wave[-1])],
    "[O I] 5577": [5568.0, 5586.0],
    "[O I] 6300/6364": [6288.0, 6374.0],
    "O2 0-1": [8580.0, 8720.0],
    "Red OH": [9300.0, 9450.0],
}


def decomposition_figure(label):
    spectrum = spectra[label]
    result = results[label]
    flux = spectrum["flux"]
    residual = flux - result.bestfit_lsf
    skyline_mask = runtime_state[label]["skyline_mask"]
    line_free = np.isfinite(residual) & ~skyline_mask
    sigma = robust_mad(residual[line_free])
    residual_scale = max(float(np.percentile(np.abs(residual), 95.0)), sigma, 0.1)
    residual_zero = (
        min(
            float(np.percentile(flux, 1.0)),
            float(np.percentile(result.bestfit_lsf, 1.0)),
        )
        - 4.5 * residual_scale
    )

    figure = go.Figure()
    figure.add_trace(
        go.Scattergl(
            x=wave,
            y=flux,
            mode="lines",
            name="Observed",
            line={"color": "black", "width": 0.8},
            hovertemplate="Observed<br>Wavelength=%{x:.2f} A<br>Flux=%{y:.6g}<extra></extra>",
        )
    )
    figure.add_trace(
        go.Scattergl(
            x=wave,
            y=result.bestfit_lsf,
            mode="lines",
            name="Best-fit full model",
            line={"color": "crimson", "width": 1.25},
            hovertemplate="Full model<br>Wavelength=%{x:.2f} A<br>Flux=%{y:.6g}<extra></extra>",
        )
    )
    for component, style in COMPONENT_STYLES.items():
        figure.add_trace(
            go.Scattergl(
                x=wave,
                y=result.components[component],
                mode="lines",
                name=style["label"],
                line={
                    "color": style["color"],
                    "dash": style["dash"],
                    "width": 0.85,
                },
                opacity=0.82,
                hovertemplate=(
                    f"{style['label']}<br>Wavelength=%{{x:.2f}} A"
                    "<br>Flux=%{y:.6g}<extra></extra>"
                ),
            )
        )

    figure.add_trace(
        go.Scattergl(
            x=wave,
            y=residual_zero + residual,
            mode="lines",
            name="Residual (shifted)",
            line={"color": "royalblue", "width": 0.9},
            customdata=residual,
            hovertemplate=(
                "Residual<br>Wavelength=%{x:.2f} A"
                "<br>Residual=%{customdata:.6g}<extra></extra>"
            ),
        )
    )
    for name, level, dash in (
        ("Residual zero", residual_zero, "dot"),
        ("+1 robust residual sigma", residual_zero + sigma, "dash"),
        ("-1 robust residual sigma", residual_zero - sigma, "dash"),
    ):
        figure.add_trace(
            go.Scattergl(
                x=wave,
                y=np.full_like(wave, level),
                mode="lines",
                name=name,
                line={"color": "gray", "dash": dash, "width": 0.7},
                hoverinfo="skip",
            )
        )

    range_buttons = [
        {
            "label": name,
            "method": "relayout",
            "args": [{"xaxis.range": bounds}],
        }
        for name, bounds in WAVELENGTH_RANGES.items()
    ]
    meta = spectrum["meta"]
    figure.update_layout(
        title=(
            f"{label} | updated PALACE | row {spectrum['row']} | "
            f"expnum {int(meta['expnum'])} | "
            f"integrated flux={spectrum['integrated_brightness']:.5g} | "
            f"{result.fit_status}, cycles "
            f"{result.lsf_state.completed_cycles}/{result.lsf_state.requested_cycles}"
        ),
        template="plotly_white",
        height=650,
        hovermode="closest",
        xaxis_title="Wavelength [Angstrom]",
        yaxis_title="Flux [input x 1e14]",
        legend={"orientation": "v", "x": 1.01, "y": 1.0},
        margin={"t": 130, "r": 250, "b": 70, "l": 80},
        updatemenus=[
            {
                "type": "buttons",
                "direction": "right",
                "buttons": range_buttons,
                "x": 0.0,
                "xanchor": "left",
                "y": 1.06,
                "yanchor": "bottom",
            }
        ],
    )
    return figure


PLOT_CONFIG = {
    "responsive": True,
    "displaylogo": False,
    "scrollZoom": True,
    "toImageButtonOptions": {"format": "png", "scale": 2},
}
```
